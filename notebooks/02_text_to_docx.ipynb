# Edu Nexus – Phase 1: Text to DOCX Writer

This notebook defines the core writer utility used across the ingestion pipeline.
Its responsibility is to convert extracted text into a clean, structured DOCX file.

All converters (PDF, PPT, OCR) will rely on this writer.


```python
# ===== IMPORTS =====
!pip install python-docx
from docx import Document
from docx.shared import Pt
from pathlib import Path
```

```python
# ===== CORE DOCX WRITER =====

def write_text_to_docx(text: str, output_path: Path, title: str = None):
    """
    Writes plain text into a structured DOCX file.

    Args:
        text (str): Extracted text content
        output_path (Path): Path to save the DOCX file
        title (str, optional): Optional document title
    """
    if not text or not text.strip():
        raise ValueError("Input text is empty. Cannot generate DOCX.")

    doc = Document()

    # Add title if provided
    if title:
        heading = doc.add_heading(title, level=1)
        heading.alignment = 1  # center

    # Add body text
    for line in text.split("\n"):
        para = doc.add_paragraph(line)
        for run in para.runs:
            run.font.size = Pt(11)

    # Ensure parent folder exists
    output_path.parent.mkdir(parents=True, exist_ok=True)

    doc.save(output_path)

    return output_path
```

```python
# ===== TEST DOCX WRITER =====

sample_text = """
Operating Systems – Introduction

An operating system is system software that manages computer hardware,
software resources, and provides common services for programs.

Key responsibilities include:
- Process management
- Memory management
- File system management
"""

test_output = Path("../../edu_nexus_db/normalized/docx/test_writer_output.docx")

write_text_to_docx(
    text=sample_text,
    output_path=test_output,
    title="Operating Systems – Basics"
)

print("Test DOCX created at:", test_output.resolve())
```
