```python
from ingest.ocr import OCR
ocr_engine = OCR()
```

```python
from pathlib import Path

repo_root = Path("..").resolve()
db_path = repo_root.parent / "edu_nexus_db"

raw_images = db_path / "raw" / "images"
intermediate_ocr = db_path / "intermediate" / "ocr_text"
normalized_docx = db_path / "normalized" / "docx"
metadata_file = db_path / "metadata" / "processing_log.json"

# create folders if missing
intermediate_ocr.mkdir(parents=True, exist_ok=True)
normalized_docx.mkdir(parents=True, exist_ok=True)

print("Paths ready ✅")

```

```python
img = list(raw_images.glob("*"))[0]

print("Testing image:", img)

text = ocr_engine.image_to_text(str(img))

print("\nExtracted Text:\n")
print(text)
```

```python
ocr_results = {}

for img in raw_images.glob("*"):
    print(f"Processing: {img.name}")
    
    extracted_text = ocr_engine.image_to_text(str(img))
    ocr_results[img.name] = extracted_text
    
    # save raw OCR output
    with open(intermediate_ocr / f"{img.stem}.txt", "w", encoding="utf-8") as f:
        f.write(extracted_text)

print("Intermediate OCR files saved ✅")
```

```python
from docx import Document

for filename, content in ocr_results.items():
    doc = Document()
    doc.add_paragraph(content)
    
    doc.save(normalized_docx / f"{Path(filename).stem}.docx")

print("DOCX files saved in normalized folder ✅")
```

```python
import json
from datetime import datetime

# load existing metadata if exists
if metadata_file.exists():
    with open(metadata_file, "r", encoding="utf-8") as f:
        metadata = json.load(f)
else:
    metadata = []

for filename in ocr_results.keys():
    entry = {
        "file_name": filename,
        "source": "raw/images",
        "processing_stage": "ocr_completed",
        "output_docx": f"normalized/docx/{Path(filename).stem}.docx",
        "timestamp": datetime.now().isoformat()
    }
    
    metadata.append(entry)

with open(metadata_file, "w", encoding="utf-8") as f:
    json.dump(metadata, f, indent=4)

print("Metadata updated ✅")
```
